19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

```python
import numpy as np

# Crio dois arrays de inteiros com 20 posições.
vetor1 = np.random.randint(1, 10, 20)
vetor2 = np.random.randint(1, 10, 20)

# Calculo o produto escalar e as normas (magnitudes) de cada vetor.
produto_escalar = np.dot(vetor1, vetor2)
norma1 = np.linalg.norm(vetor1)
norma2 = np.linalg.norm(vetor2)

# O cosseno do ângulo é o produto escalar dividido pela multiplicação das normas.
cos_theta = produto_escalar / (norma1 * norma2)

# Utilizo arccos para achar o ângulo em radianos e depois converto para graus. (np.clip evita pequenos erros de ponto flutuante).
angulo_radianos = np.arccos(np.clip(cos_theta, -1.0, 1.0))
angulo_graus = np.degrees(angulo_radianos)

# Exibo o ângulo calculado em graus.
print("Ângulo entre os vetores (em graus):", angulo_graus)
```

1. Calcule a integral da função f(x) = x^2 sin(x) no intervalo de 0 a π.

```python
import numpy as np
from scipy.integrate import quad

# Defino a função f(x) que será integrada.
def f(x):
    return (x**2) * np.sin(x)

# Uso a função quad para calcular a integral definida de 0 até o valor de pi (np.pi).
resultado, erro = quad(f, 0, np.pi)

# Exibo o resultado da integral e a estimativa de erro.
print("Resultado da integral:", resultado)
print("Erro estimado:", erro)
```

2. Resolva a equação diferencial dy/dx = x - y, com y(0) = 1, no intervalo de 0 a 10.

```python
import numpy as np
from scipy.integrate import solve_ivp

# Defino a função da equação diferencial, que recebe t (ou x) e y.
def dy_dx(x, y):
    return x - y

# Uso solve_ivp para resolver a equação. Passo a função, o intervalo (0 a 10) e a condição inicial y(0) = 1.
solucao = solve_ivp(dy_dx, [0, 10], [1])

# Exibo os valores de x (t) e y calculados pelo solucionador.
print("Valores de x calculados:", solucao.t)
print("Valores de y calculados:", solucao.y[0])
```

3. Encontre as raízes da função f(x) = x^4 - 3x^3 + 2 utilizando SciPy.

```python
from scipy.optimize import fsolve

# Defino a função matemática.
def f(x):
    return x**4 - 3*x**3 + 2

# Uso fsolve e passo alguns "chutes" iniciais (guesses) para encontrar diferentes raízes da função.
raizes = fsolve(f, [-1, 1, 3])

# Exibo as raízes encontradas (arredondando para facilitar a leitura).
print("Raízes da função:", np.round(raizes, 4))
```

4. Use a função de otimização de SciPy para maximizar f(x) = -x^2 + 4x + 1.

```python
from scipy.optimize import minimize

# Para maximizar uma função no SciPy, eu minimizo a sua versão negativa.
# Então, defino a função como a negativa da original: -(-x^2 + 4x + 1) = x^2 - 4x - 1.
def f_negativa(x):
    return x**2 - 4*x - 1

# Uso a função minimize com um chute inicial (x = 0).
resultado = minimize(f_negativa, 0)

# Exibo o valor de x que maximiza a função original.
print("O valor de x que maximiza a função é:", resultado.x[0])
```

12. Verifique quais colunas de um DataFrame contêm valores duplicados.

```python
import pandas as pd

# Crio um DataFrame onde a coluna 'A' tem valores duplicados e a 'B' não.
df = pd.DataFrame({'A': [1, 1, 3], 'B': [4, 5, 6]})

# Aplico uma função anônima em cada coluna (x) para checar se x.duplicated().any() é verdadeiro.
colunas_com_duplicados = df.columns[df.apply(lambda x: x.duplicated().any())]

# Exibo os nomes das colunas que possuem valores repetidos.
print("Colunas com valores duplicados:", list(colunas_com_duplicados))
```

6. Plote dois gráficos de linha em subplots diferentes: y1 = e^x e y2 = log(x).

8. Plote um gráfico de linha com intervalos de confiança (sombramento) para a função y = sin(x).

```python
import numpy as np
import matplotlib.pyplot as plt

# Crio o intervalo de x e calculo o seno de x.
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Crio um valor constante para simular o "erro" ou intervalo de confiança.
erro = 0.2

# Ploto a linha principal.
plt.plot(x, y, color='blue')

# Uso fill_between para criar o sombreado entre y-erro e y+erro.
plt.fill_between(x, y - erro, y + erro, color='blue', alpha=0.2)

# Exibo o resultado.
plt.show()
```
